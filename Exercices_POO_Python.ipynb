```python
# Exrcice 1 :
from abc import ABC , abstractmethod 
from datetime import datetime
class Employe (ABC) :
    SMIG = 2500
    def __init__(self, nom , prenom , matricule , annee_naissance ):
        self.nom = nom
        self.prenom = prenom
        self.matricule = matricule
        self.annee_naissance = annee_naissance
    def __str__(self) -> str:
        return f"Bonjour {self.nom} {self.prenom} , Votre matricule est : {self.matricule} , tu as nai en {self.annee_naissance}"
    @abstractmethod 
    def get_salaire (self)->float :
        pass
class Ouvrier(Employe) :
    def __init__(self, nom, prenom, matricule, annee_naissance , annee_entree):
        super().__init__(nom, prenom, matricule, annee_naissance)
        self.annee_entree = annee_entree
    def get_salaire(self)->float:
        return min(self.SMIG + (datetime.now().year - self.annee_entree) * 100 , self.SMIG * 2)
class Cadre(Employe) :
    def __init__(self, nom, prenom, matricule, annee_naissance , indice):
        super().__init__(nom, prenom, matricule, annee_naissance)
        self.indice = indice
    def get_salaire(self)->float:
        return {1: 13000 , 2:15000 , 3:17000 }[self.indice]
class Patron(Employe) :
    chiffre_affaire = 1000000
    def __init__(self, nom, prenom, matricule, annee_naissance):
        super().__init__(nom, prenom, matricule, annee_naissance)
    def get_salaire(self)->float:
        return self.chiffre_affaire * 0.1
cadre = Cadre('Elkerymy' , 'Mohamed ' , 20050920 , 2005 , 3)
print(cadre)
print(cadre.get_salaire())
```
