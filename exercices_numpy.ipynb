## Exercice 2 : Extraction par masque booléen (1D)

**Objectif :**
Soit un vecteur 1D de 15 nombres aléatoires entre 0 et 50. Extrayez tous les éléments strictement supérieurs à 25 **ET** pairs.

```python
# VOTRE CODE ICI
v = np.random.randint(0 , 50 , 15)
print(v)

result = v[(v > 25) & (v % 2 == 0)]
print(result)
```

## Exercice 3 : Inversion et inversion par colonnes (1D et 2D)

2. Soit une matrice $4 \times 4$, inversez l'ordre de ses colonnes (la dernière colonne devient la première).

## Exercice 4 : Statistique par axe (2D)

**Objectif :**
Soit une matrice $5 \times 3$ d'entiers aléatoires entre 1 et 100.
1. Calculez la moyenne globale de la matrice.
2. Calculez la somme des éléments pour chaque **colonne**.
3. Obtenez la valeur maximale pour chaque **ligne**.

```python
# VOTRE CODE ICI
matrice = np.random.randint(1 , 100 , 15).reshape(5 , 3)
print(matrice)
print(np.mean(matrice))
print(matrice.sum(axis=0))
print(matrice.max(axis=1))
```

## Exercice 5 : Remplacement des valeurs manquantes (1D / 2D)

**Objectif :**
Soit le tableau : `arr = np.array([1.0, 2.0, np.nan, 4.0, np.nan, 6.0])`.
Remplacez toutes les valeurs `NaN` par la moyenne des valeurs non manquantes du tableau.

```python
# VOTRE CODE ICI
arr = np.array([1.0 , 2.0 , np.nan , 4.0 , np.nan , 6.0])
avg = np.nanmean(arr)
arr[np.isnan(arr) ]= avg
print(arr)
```

## Exercice 6 : Concaténation et empilement (2D)

**Objectif :**
Soit deux matrices $A$ et $B$ de dimensions $3 \times 2$.
1. Empilez-les verticalement pour obtenir une matrice $6 \times 2$.
2. Empilez-les horizontalement pour obtenir une matrice $3 \times 4$.

```python
# VOTRE CODE ICI
A = np.random.randint(1 , 100 , 6).reshape(3 , 2)
B = np.random.randint(1 , 100 , 6).reshape(3 , 2)
v = np.vstack((A , B))
h = np.hstack((A, B))
print(v)
print(h)
```

## Exercice 7 : Recherche d'indices spécifiques (`np.where`) (1D)

**Objectif :**
Soit un vecteur 1D de 12 valeurs réelles. Récupérez les **indices** (et non les valeurs) de tous les éléments compris entre 10 et 20 inclus.

```python
# VOTRE CODE ICI
v = np.random.randint(1 , 30 , 12)
indices = np.where((v >= 10) & (v <= 20))[0]
print(indices)
```

## Exercice 9 : Masquage conditionnel & Remplacement rapide (1D)

**Objectif :**
Créez un vecteur 1D de 20 entiers aléatoires entre -10 et 10. Sans utiliser de boucle for ni d'instructions if, remplacez tous les nombres négatifs impairs par leur carré, et tous les nombres positifs pairs par 0.

```python
# VOTRE CODE ICI
vecteur1D = np.random.randint(-10 , 10 , 20)
print(vecteur1D)
vecteur1D[vecteur1D > 0] = 0
vecteur1D[vecteur1D < 0] = vecteur1D[vecteur1D < 0] ** 2
print(vecteur1D)
```

## Exercice 10 : Extraction par glissement / Fenêtres glissantes (1D)

**Objectif :**
Soit un vecteur 1D $V = [1, 2, 3, 4, 5, 6, 7, 8]$. Générez une matrice 2D représentant toutes les sous-fenêtres glissantes de taille 3 avec un pas (stride) de 1.Résultat attendu :$$\begin{bmatrix} 1 & 2 & 3 \\ 2 & 3 & 4 \\ 3 & 4 & 5 \\ \dots & \dots & \dots \end{bmatrix}$$

```python
# VOTRE CODE ICI
v = np.arange(1 , 9)
vs = np.lib.stride_tricks.sliding_window_view(v, window_shape=3)
print(vs)
```

## Exercice 11 : Extraction des k plus grands éléments par ligne (2D)

**Objectif :**
Soit une matrice 2D de dimensions $6 \times 8$ remplie d'entiers aléatoires. Extrayez, pour chaque ligne, les 3 plus grandes valeurs et leurs indices respectifs, triés par ordre décroissant.

```python
# VOTRE CODE ICI
matrice = np.random.randint(1 , 100 , 48).reshape(6 , 8)
t3indices = np.argsort(matrice , axis=1)[:,-3 :][::-1]
t3valeur = np.take_along_axis(matrice , t3indices , axis=1)
print('indices : -----' , '\n' , t3indices) 
print('valeur : -----' , '\n' , t3valeur) 
```
